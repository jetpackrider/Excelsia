# Excelsia on a free Kaggle GPU

**To run:** set the two numbers in the box below, then click **Save Version**, choose **Save & Run All**, and save. It runs in the background on Kaggle's machines. You can close the tab and come back later.

- `CHATS`: how many practice chats Gemma writes.
- `ENGLISH_MINUTES`: how long to train her English. `0` means only write chats, which is the quick way to check the chats are any good.

When it finishes, open the finished version. The printed log ends with a sample of chats (and a test conversation, if it trained). The files are under **Output**, in `excelsia_out`.

```python
REPO = "https://github.com/jetpackrider/Excelsia"
CHATS = 300
ENGLISH_MINUTES = 0

%cd /kaggle/working
!rm -rf /tmp/smol-gpt && git clone -q --depth 1 $REPO /tmp/smol-gpt
%cd /tmp/smol-gpt
!EXCELSIA_CHATS=$CHATS EXCELSIA_MINUTES=$ENGLISH_MINUTES bash scripts/kaggle.sh all
```
